````python
import yfinance as yf 
````

````python
# download the data

df = yf.download("AAPL",
                start="2011-01-01",
                end="2021-12-31",
                progress=False)
````

````python
# inspect the downloaded data:
print(f'Download {len(df)} rows of data.')
df.head(5)
````

#### Additional Features

* We can download information for multiple tickers at once by providing a list of tickers `(["AAPL", "MSFT"])` or multiple tickers as a string `("AAPL MSFT")`.
* We can set `auto_adjust=True` to download only the adjusted prices.
* We can additionally download dividends and stock splits by setting `actions='inline'`. Those
actions can also be used to manually adjust the prices or for other analyses.
* Specifying `progress=False` disables the progress bar.
* The `interval` argument can be used to download data in different frequencies. We could also
download intraday data as long as the requested period is shorter than 60 days.

#### Using multi_level_index=False to sort the headers

Currently, `multi_level_index` is by default `True`, so you get this when you download data:

```text
Price       Close    High    Low    Open    Volume
Ticker      AAPL     AAPL    AAPL   AAPL    AAPL
Date
```
When you use:

```python
multi_level_index=False
```
You should look more like this:

```text
Date          Open       High       Low        Close      Volume
2011-01-03    ...
2011-01-04    ...
2011-01-05    ...
```


````python
df = yf.download(
    "AAPL", # ticker
    start="2011-01-01",
    end="2021-12-31",
    progress=False, # disable displaying the progress bar for downloading data
    multi_level_index=False, # fixes the headers
)

df.head()
````

### What does `prepost=True` actually mean?

For a US stock or ETF such as AAPL or QQQ, there are three trading sessions:

```text
Pre-market      Regular market        After-hours
4:00–9:30 ET    9:30–16:00 ET         16:00–20:00 ET
      │                │                     │
      └──────── extended trading ────────────┘
```

Nasdaq describes pre-market as 4:00–9:30 ET and after-hours as 16:00–20:00 ET. These are trading sessions before and after the main exchange session.

so:
```python
prepost=False
```
means:

> Give me the normal/regular trading session

whereas:
```text 
prepost=True
```
means:
> Include the pre-market and after-hours trades as well

#### The QuantConnect comparison
