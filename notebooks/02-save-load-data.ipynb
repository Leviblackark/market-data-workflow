````python
# TODO 1: import whats needed
from pathlib import Path
import pandas as pd
import yfinance as yf
````

````python
# TODO 2: Create a clean DataFrame

ICKER = "QQQ"

df = yf.download(
    TICKER,
    period="5d",
    interval="1m", 
    auto_adjust=True,
    prepost=False, 
    multi_level_index=False,
    progress=False
)

# access the columns, treat those columns as a string and convert to lower case
df.columns = df.columns.str.lower()
# reorder the order in the df 
df = df[["open", "high", "low", "close", "volume"]]

# convert the index to UTC timezone
df.index = df.index.tz_convert("UTC")

# Quantconnect works with the bar endtime, so we need to shift the index by 1 minute
bar_length = pd.Timedelta(minutes=1)
df.index = df.index + bar_length

# match the index name to quantconnects df
df.index.name = "time"
df = df.sort_index()
````

````python
# TODO 3: Create the file path 

data_folder = Path("../data")
````

````python
# add the file name to the file path
file_path = data_folder / "qqq_1min.csv"
````

````python
# TODO 4: save the dataframe to a csv

df.to_csv(file_path)
````
